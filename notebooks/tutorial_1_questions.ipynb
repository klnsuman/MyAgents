```python
# Cell 2: Quick Reference - TypedDict
"""
CONCEPT: TypedDict

What is it?
-----------
A way to define the structure/schema of a dictionary with type hints.

Example:
--------
"""

from typing import TypedDict

# Define schema
class Person(TypedDict):
    name: str
    age: int
    email: str

# Use it
person: Person = {
    "name": "Alice",
    "age": 30,
    "email": "alice@example.com"
}

print(person)
print(f"Type: {type(person)}")  # Still a regular dict!

# What if I make a typo?
# person_wrong = {
#     "nam": "Bob",  # Typo! IDE will warn you
#     "age": 25
# }

# What if I use wrong type?
# person_wrong2 = {
#     "name": 123,  # Wrong type! IDE will warn you
#     "age": 25
# }

print("\n✓ TypedDict provides type safety for dictionaries")
```
